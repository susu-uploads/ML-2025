### Импорт библиотек

```python
import matplotlib.pyplot as plt
import mglearn
from scipy.io import loadmat
from sklearn import svm
```

# Часть 1: Линейный SVM для линейно разделимых классов

### Загрузка данных

```python
data = loadmat('data/ex4data1.mat')
data.keys()
```

### Инициализация векторов

```python
X = data["X"]
y = data["y"].ravel()
```

### Визуализация данных

```python
plt.figure(figsize=(8, 6))
plt.scatter(X[:,0][y == 1], X[:,1][y == 1], marker="^", s=100, facecolor='orange', edgecolors='black')
plt.scatter(X[:,0][y == 0], X[:,1][y == 0], marker="o", s=100, facecolor='blue', edgecolors='black')
plt.show()
```

### Функция обучения SVC классификатора

```python
def fit_svc_classifier(X, y, C, loss, max_iter) -> svm.LinearSVC:
    svc_classifier = svm.LinearSVC(C=C, loss=loss, max_iter=max_iter)
    svc_classifier.fit(X, y)
    print(f'Score: {svc_classifier.score(X, y)}')
    return svc_classifier
```

### Функция отображения графика разделения выборки SVC классификатором

```python
def svc_classifier_plot(svc_classifier, X, y, eps=.5):
    mglearn.plots.plot_2d_separator(svc_classifier, X, eps=eps)
    mglearn.discrete_scatter(X[:, 0], X[:, 1], y)
```

### Обучение SVC классификатора при `C = 1`

```python
svc_classifier = fit_svc_classifier(X, y, 1, 'hinge', 10000)
```

### Отображение графика разделения тренировочной выборки с помощью SVC классификатора

```python
svc_classifier_plot(svc_classifier, X, y)
```

### Обучение SVC классификатора при `C = 2000`

```python
svc_classifier = fit_svc_classifier(X, y, 2000, 'hinge', 10000)
```

### Отображение графика разделения тренировочной выборки с помощью SVC классификатора

```python
svc_classifier_plot(svc_classifier, X, y)
```

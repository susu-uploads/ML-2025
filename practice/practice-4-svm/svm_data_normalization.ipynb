### Импорт библиотек

```python
import matplotlib.pyplot as plt
from sklearn import svm
```

# Часть 4: Влияние предварительной нормализации данных на работу SVM

### Загрузка данных

```python
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()
```

### Разбиение датасета на тренировочную и тестовую выборки

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target)
```

### Обучение SVC классификатора на ненормализованной тренировочной выборке

```python
svc_classifier = svm.SVC()
svc_classifier.fit(X_train, y_train)
```

### Качество обучения SVC классификатора на ненормализованной тренировочной и тестовой выборках

```python
print("Правильность на обучающем наборе: {:.2f}".format(svc_classifier.score(X_train, y_train)))
print("Правильность на тестовом наборе: {:.2f}".format(svc_classifier.score(X_test, y_test)))
```

### Оценка необходимости нормализации данных

```python
plt.plot(X_train.min(axis=0), 'o', label="min")
plt.plot(X_train.max(axis=0), '^', label="max")
plt.legend(loc=4)
plt.xlabel("Индекс признака")
plt.ylabel("Величина признака")
plt.yscale("log")
```

### Нормализация данных

```python
min_value = X_train.min(axis=0)
max_value = X_train.max(axis=0)
X_train_norm = (X_train - min_value) / (max_value - min_value)
X_test_norm = (X_test - min_value) / (max_value - min_value)
```

### Различие между нормализованными и ненормализованными данными

```python
plt.plot(X_train_norm.min(axis=0), 'o', label="min")
plt.plot(X_train_norm.max(axis=0), '^', label="max")
plt.legend(loc=4)
plt.xlabel("Индекс признака")
plt.ylabel("Величина признака")
```

### Обучение SVC классификатора на нормализованной тренировочной выборке

```python
svc_classifier = svm.SVC()
svc_classifier.fit(X_train_norm, y_train)
```

### Качество обучения SVC классификатора на нормализованной тренировочной и тестовой выборках

```python
print("Правильность на обучающем наборе: {:.2f}".format(svc_classifier.score(X_train_norm, y_train)))
print("Правильность на тестовом наборе: {:.2f}".format(svc_classifier.score(X_test_norm, y_test)))
```

### Обучение SVC классификатора с дополнительной настройкой параметров на нормализованной тренировочной выборке

```python
svc_classifier = svm.SVC(kernel='rbf', C=2, gamma=2)
svc_classifier.fit(X_train_norm, y_train)
```

### Качество обучения SVC классификатора с дополнительной настройкой параметров на нормализованной тренировочной и тестовой выборках

```python
print("Правильность на обучающем наборе: {:.2f}".format(svc_classifier.score(X_train_norm, y_train)))
print("Правильность на тестовом наборе: {:.2f}".format(svc_classifier.score(X_test_norm, y_test)))
```
